## Task 2: Data Cleaning

In this section, we clean the dataset for recommendation modeling:
- Remove non-informative index column(s)
- Handle missing values in key identity fields
- Resolve duplicate song records with a clear strategy
- Standardize data types
- Save a cleaned dataset for Task 3

```python
# Create working copy for cleaning

df_clean = df.copy()

rows_before = len(df_clean)
cols_before = len(df_clean.columns)

print('Before cleaning:')
print(f'Rows: {rows_before}')
print(f'Columns: {cols_before}')
print(f'Total missing values: {int(df_clean.isna().sum().sum())}')
print(f'Duplicate track_id: {int(df_clean["track_id"].duplicated().sum()) if "track_id" in df_clean.columns else "N/A"}')
```

```python
# Step 1: Remove non-informative index-like columns
for col in ['Unnamed: 0']:
    if col in df_clean.columns:
        df_clean = df_clean.drop(columns=[col])

# Step 2: Drop rows with missing values in key identity fields
key_identity_cols = [c for c in ['track_id', 'track_name', 'artists', 'album_name', 'track_genre'] if c in df_clean.columns]
missing_before_drop = int(df_clean[key_identity_cols].isna().sum().sum()) if key_identity_cols else 0

df_clean = df_clean.dropna(subset=key_identity_cols)

# Step 3: Remove duplicate track_id entries (keep first)
dup_track_id_before = int(df_clean['track_id'].duplicated().sum()) if 'track_id' in df_clean.columns else 0
if 'track_id' in df_clean.columns:
    df_clean = df_clean.drop_duplicates(subset=['track_id'], keep='first')

# Step 4: Remove duplicate song-artist-album combinations (keep first)
dup_triplet_cols = [c for c in ['track_name', 'artists', 'album_name'] if c in df_clean.columns]
dup_triplet_before = int(df_clean.duplicated(subset=dup_triplet_cols).sum()) if dup_triplet_cols else 0
if dup_triplet_cols:
    df_clean = df_clean.drop_duplicates(subset=dup_triplet_cols, keep='first')

print('Cleaning actions completed.')
print(f'Missing (key identity) before drop: {missing_before_drop}')
print(f'Duplicate track_id before drop: {dup_track_id_before}')
print(f'Duplicate song-artist-album before drop: {dup_triplet_before}')
```

```python
# Step 5: Standardize data types for modeling

if 'explicit' in df_clean.columns:
    # Convert bool/object to clean numeric flag: 0 or 1
    explicit_map = {True: 1, False: 0, 'True': 1, 'False': 0}
    df_clean['explicit'] = df_clean['explicit'].map(explicit_map).fillna(0).astype(int)

# Keep integer semantic columns as numeric where applicable
for int_col in ['key', 'mode', 'time_signature', 'popularity', 'duration_ms']:
    if int_col in df_clean.columns:
        df_clean[int_col] = pd.to_numeric(df_clean[int_col], errors='coerce')

print('Data types standardized for key columns.')
```

```python
# Post-cleaning quality checks

rows_after = len(df_clean)
cols_after = len(df_clean.columns)

quality_report = {
    'rows_before': rows_before,
    'rows_after': rows_after,
    'rows_removed': rows_before - rows_after,
    'cols_before': cols_before,
    'cols_after': cols_after,
    'total_missing_after': int(df_clean.isna().sum().sum()),
    'duplicate_rows_after': int(df_clean.duplicated().sum()),
    'duplicate_track_id_after': int(df_clean['track_id'].duplicated().sum()) if 'track_id' in df_clean.columns else None,
    'duplicate_song_artist_album_after': int(df_clean.duplicated(subset=dup_triplet_cols).sum()) if dup_triplet_cols else None,
}

print('TASK 2 CLEANING REPORT')
print('-' * 60)
for k, v in quality_report.items():
    print(f'{k}: {v}')

print('\nCleaned dataset preview:')
display(df_clean.head(5))
```

## Task 4: Build the Recommendation Engine (Content-Based)

This section builds the recommender using the scaled audio-feature vectors:
- Train a nearest-neighbor retrieval model (cosine distance)
- Create helper functions for exact/partial song search
- Return top-N similar songs with similarity scores
- Demonstrate recommendations on sample songs

```python
from sklearn.neighbors import NearestNeighbors

# Ensure model inputs exist (if notebook was restarted)
if 'df_clean' not in globals():
    df_clean = pd.read_csv('dataset_cleaned.csv')

if 'X_scaled' not in globals() or 'feature_cols' not in globals():
    candidate_features = [
        'danceability', 'energy', 'tempo', 'valence', 'acousticness',
        'instrumentalness', 'liveness', 'speechiness', 'loudness',
        'duration_ms', 'popularity', 'explicit', 'key', 'mode', 'time_signature'
    ]
    feature_cols = [c for c in candidate_features if c in df_clean.columns]
    X = df_clean[feature_cols].apply(pd.to_numeric, errors='coerce').fillna(df_clean[feature_cols].median(numeric_only=True))
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

print(f'Recommender input matrix shape: {X_scaled.shape}')
print(f'Feature count: {len(feature_cols)}')
```

```python
# Fit nearest-neighbor model using cosine distance
knn_model = NearestNeighbors(metric='cosine', algorithm='brute')
knn_model.fit(X_scaled)

print('NearestNeighbors model trained successfully.')
```

```python
# Build helper lookup columns for robust search
search_df = df_clean.copy()
search_df['track_name_lower'] = search_df['track_name'].astype(str).str.lower().str.strip()
search_df['artists_lower'] = search_df['artists'].astype(str).str.lower().str.strip()

# Optional disambiguation support: same song name by different artists
name_to_indices = search_df.groupby('track_name_lower').indices

print(f'Unique searchable song names: {len(name_to_indices)}')
```

```python
def find_song_candidates(song_name, max_candidates=10):
    """Return candidate rows for exact/partial song-name matches."""
    query = str(song_name).lower().strip()

    exact_mask = search_df['track_name_lower'] == query
    exact_matches = search_df[exact_mask]

    if not exact_matches.empty:
        return exact_matches[['track_name', 'artists', 'track_genre']].head(max_candidates)

    partial_mask = search_df['track_name_lower'].str.contains(query, na=False)
    partial_matches = search_df[partial_mask]

    return partial_matches[['track_name', 'artists', 'track_genre']].drop_duplicates().head(max_candidates)


def recommend_songs(song_name, top_n=10, artist_name=None):
    """
    Recommend songs similar to a given song using cosine-nearest neighbors.
    Optional artist_name helps disambiguate duplicates.
    """
    query = str(song_name).lower().strip()
    candidate_rows = search_df[search_df['track_name_lower'] == query]

    if candidate_rows.empty:
        partial_mask = search_df['track_name_lower'].str.contains(query, na=False)
        partial_candidates = search_df[partial_mask][['track_name', 'artists', 'track_genre']].drop_duplicates().head(10)
        return {
            'status': 'not_found',
            'message': f'Song not found: {song_name}',
            'suggestions': partial_candidates
        }

    if artist_name is not None:
        artist_query = str(artist_name).lower().strip()
        artist_filtered = candidate_rows[candidate_rows['artists_lower'].str.contains(artist_query, na=False)]
        if not artist_filtered.empty:
            candidate_rows = artist_filtered

    query_index = candidate_rows.index[0]

    distances, indices = knn_model.kneighbors(X_scaled[query_index].reshape(1, -1), n_neighbors=top_n + 1)

    # Exclude the query song itself, convert cosine distance to similarity
    rec_indices = [idx for idx in indices.flatten() if idx != query_index][:top_n]
    rec_distances = []
    for idx, dist in zip(indices.flatten(), distances.flatten()):
        if idx != query_index:
            rec_distances.append(dist)
        if len(rec_distances) == top_n:
            break

    rec_df = df_clean.iloc[rec_indices][['track_name', 'artists', 'track_genre', 'popularity']].copy()
    rec_df['cosine_distance'] = rec_distances
    rec_df['similarity_score'] = 1 - rec_df['cosine_distance']
    rec_df = rec_df.sort_values(by='similarity_score', ascending=False).reset_index(drop=True)

    return {
        'status': 'ok',
        'query_song': df_clean.loc[query_index, ['track_name', 'artists', 'track_genre']].to_dict(),
        'recommendations': rec_df
    }
```

## Task 5: Evaluate and Analyze Recommendation Results

This section evaluates recommendation quality using practical proxy metrics (since no user-rating ground truth is available):
- Average similarity score
- Query-genre match rate in top recommendations
- Artist diversity in recommendations
- Popularity gap between query and recommended songs

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')

# Queries for evaluation (picked from songs known to exist in dataset)
eval_queries = [
    {'song': "I'm Yours", 'artist': 'Jason Mraz'},
    {'song': 'Say Something', 'artist': None},
    {'song': 'Brave', 'artist': 'Sara Bareilles'},
    {'song': 'Boston', 'artist': 'Augustana'},
    {'song': 'Hold On', 'artist': 'Chord Overstreet'},
]

print('Evaluation query set:')
for q in eval_queries:
    print(q)
```

```python
# Run evaluation and collect metrics per query

eval_rows = []
all_recs_long = []

for q in eval_queries:
    res = recommend_songs(q['song'], top_n=10, artist_name=q['artist'])

    if res['status'] != 'ok':
        eval_rows.append({
            'query_song': q['song'],
            'query_artist': q['artist'],
            'status': 'not_found',
            'avg_similarity': np.nan,
            'genre_match_rate': np.nan,
            'unique_artist_count': np.nan,
            'avg_popularity_gap': np.nan
        })
        continue

    rec_df = res['recommendations'].copy()
    query_info = res['query_song']

    query_genre = query_info.get('track_genre', None)
    query_artist = query_info.get('artists', None)

    genre_match_rate = (rec_df['track_genre'] == query_genre).mean() if query_genre is not None else np.nan
    unique_artist_count = rec_df['artists'].nunique()

    query_pop_rows = df_clean[
        (df_clean['track_name'] == query_info.get('track_name')) &
        (df_clean['artists'] == query_info.get('artists'))
    ]
    query_popularity = query_pop_rows['popularity'].iloc[0] if not query_pop_rows.empty else np.nan
    avg_popularity_gap = (rec_df['popularity'] - query_popularity).abs().mean() if pd.notna(query_popularity) else np.nan

    eval_rows.append({
        'query_song': query_info.get('track_name'),
        'query_artist': query_artist,
        'status': 'ok',
        'avg_similarity': rec_df['similarity_score'].mean(),
        'genre_match_rate': genre_match_rate,
        'unique_artist_count': unique_artist_count,
        'avg_popularity_gap': avg_popularity_gap
    })

    temp = rec_df.copy()
    temp['query_song'] = query_info.get('track_name')
    all_recs_long.append(temp)

eval_df = pd.DataFrame(eval_rows)
all_recs_df = pd.concat(all_recs_long, ignore_index=True) if all_recs_long else pd.DataFrame()

print('Per-query evaluation metrics:')
display(eval_df)
```

```python
# Overall summary metrics
valid_eval = eval_df[eval_df['status'] == 'ok'].copy()

overall_metrics = {
    'queries_evaluated': int(len(valid_eval)),
    'overall_avg_similarity': float(valid_eval['avg_similarity'].mean()) if not valid_eval.empty else np.nan,
    'overall_avg_genre_match_rate': float(valid_eval['genre_match_rate'].mean()) if not valid_eval.empty else np.nan,
    'overall_avg_unique_artist_count': float(valid_eval['unique_artist_count'].mean()) if not valid_eval.empty else np.nan,
    'overall_avg_popularity_gap': float(valid_eval['avg_popularity_gap'].mean()) if not valid_eval.empty else np.nan,
}

print('TASK 5 OVERALL METRICS')
print('-' * 60)
for k, v in overall_metrics.items():
    print(f'{k}: {v}')
```
